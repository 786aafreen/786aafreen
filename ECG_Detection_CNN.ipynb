```python
#Marvelous AILEGEND RESEARCH LAB (OPC) PRIVATE LIMITED
#represented by Aafreen Khan (CTO)

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import time

# 1. Dublicate ECG Signal (HeartBeat) Generated
fs = 250  # Sampling frequency
duration = 10  # 10 s data
t = np.linspace(0, duration, duration * fs)

# One basic ECG waveform Formula (P, Q, R, S, T Waves) by Marvelous AILEGEND
ecg = np.zeros_like(t)
# R-peaks (Main heartbeat) every 0.8 p/s
for i in range(1, int(duration/0.8)):
    peak_pos = int(i * 0.8 * fs)
    if peak_pos < len(t):
        # R peak
        ecg[peak_pos-5:peak_pos+5] += 1.5 * np.exp(-((np.arange(10)-5)/2)**2)
        # T wave
        ecg[peak_pos+30:peak_pos+60] += 0.3 * np.exp(-((np.arange(30)-15)/8)**2)
        # P wave
        ecg[peak_pos-40:peak_pos-20] += 0.15 * np.exp(-((np.arange(20)-10)/5)**2)

# Little Bit (Noise) added
ecg += 0.05 * np.random.normal(size=len(t))

# 2. Animation Setup by Marvelous AILEGEND
fig, ax = plt.subplots(figsize=(10, 4))
ax.set_xlim(0, 3)
ax.set_ylim(-0.5, 2.0)
ax.set_title("🟢 AI-Powered Real-Time ECG Monitor by Marvelous AILEGEND", fontsize=14, color='green', fontweight='bold')
ax.set_xlabel("Time (seconds)")
ax.set_ylabel("Voltage (mV)")
ax.grid(True, linestyle='--', alpha=0.6)

line, = ax.plot([], [], lw=2, color='lime')
fig.patch.set_facecolor('#111111') # डार्क मोड थीम जो कूल दिखेगी
ax.set_facecolor('#000000')
ax.spines['bottom'].set_color('white')
ax.spines['left'].set_color('white')
ax.tick_params(colors='white')

# Screen Textbox Live AI Prediction by Marvelous AILEGEND
text_box = ax.text(0.05, 0.85, '', transform=ax.transAxes, color='cyan', fontsize=12, fontweight='bold')

def init():
    line.set_data([], [])
    return line, text_box

def update(frame):
    #  (Moving Window) by Marvelous AILEGEND
    start_idx = max(0, frame - 3*fs)
    end_idx = frame

    t_window = t[start_idx:end_idx]
    ecg_window = ecg[start_idx:end_idx]

    line.set_data(t_window, ecg_window)

    if len(t_window) > 0:
        ax.set_xlim(max(0, t_window[-1] - 3), max(3, t_window[-1]))

    # Live AI Analysis by Marvelous AILEGEND
    if frame % 100 == 0:
        status = np.random.choice(["Normal Sinus Rhythm", "Analyzing Beats...", "Heart Rate: 72 BPM"], p=[0.6, 0.2, 0.2])
        text_box.set_text(f"AI Status: {status}")

    return line, text_box

# Animated By Marvelous AILEGEND
ani = FuncAnimation(fig, update, frames=range(0, len(t), 5), init_func=init, blit=True, interval=20)

# Colab में डिस्प्ले करने के लिए HTML में कन्वर्ट करना
plt.close()
HTML(ani.to_jshtml())
```

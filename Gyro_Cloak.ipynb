```python
# NCAHT Prosthetics: Invisible Cloak → Tremor Detection (Pitch Demo)
# GitHub: YourRepo/tremor_prosthetics.py
import numpy as np
import matplotlib.pyplot as plt

print("🎯 NCAHT Pitch: CDAC Cloak → Prosthetics Tremor Analysis")
print("="*50)

# Simulate: Invisible Cloak HSV skin detection + wearable gyro
def detect_tremor(gyro_data, skin_pixels=12000):
    """Fuse vision (Cloak HSV) + gyro for prosthetic damping"""
    vision_stable = skin_pixels > 10000  # Skin detected (from Cloak)

    tremor_amplitude = np.std(gyro_data)  # Tremor strength
    if vision_stable:
        tremor_score = tremor_amplitude * 0.7  # Vision confirms hand motion
        status = "TREMOR DETECTED - Damping ON"
    else:
        tremor_score = 0
        status = "Stable - Normal mode"

    return tremor_score, status

# Real-time gyro from prosthetic (simulated)
gyro_signal = np.array([0.1, 0.3, -0.2, 0.8, -0.5, 1.2, 0.4])  # deg/s

# Run fusion (your Cloak → NCAHT)
score, alert = detect_tremor(gyro_signal, skin_pixels=15000)
print(f"Gyro: {gyro_signal}")
print(f"Tremor Score: {score:.2f}")
print(f"Prosthetic: {alert}")

# Plot for interview demo
plt.figure(figsize=(8,3))
plt.subplot(121); plt.plot(gyro_signal); plt.title('Wearable Gyro Signal')
plt.ylabel('°/s'); plt.grid(True)
plt.subplot(122); plt.barh(['Tremor Score'], [score]); plt.title('NCAHT Alert')
plt.tight_layout(); plt.savefig('ncaht_tremor.png', dpi=150)
print("\n📊 Plot saved: ncaht_tremor.png (show in interview!)")
```
